## **1. `print()` 함수와 데이터 출력**

컴퓨터는 우리가 시키는 대로만 움직입니다. 화면에 무언가를 보여달라고 명령할 때 사용하는 것이 바로 `print()` 함수입니다.

```python
print("여기에 보여줄 내용")
```

`print`는 **함수 이름**이고, 괄호 `( )` 안에 들어가는 내용을 **인자(argument)** 라고 부릅니다. "print 함수에게 이 인자를 화면에 출력하라고 시킨다"고 이해하면 됩니다.

### 📌 따옴표가 데이터를 다르게 만든다
* **따옴표(`" "` 또는 `' '`)가 있을 때:** 컴퓨터는 이것을 계산할 필요가 없는 **글자(문자열, string)** 로 인식하고 안에 적힌 그대로 출력합니다.
* **따옴표가 없을 때:** 컴퓨터는 이것을 **숫자나 계산식**으로 인식해서 먼저 계산한 뒤 그 결과를 출력합니다.

### 📌 쉼표(`,`)로 여러 데이터 나열하기
`print()` 안에 쉼표로 여러 데이터를 나열하면, 각 데이터 사이에 **띄어쓰기가 자동으로** 들어갑니다.

### 📌 문자열도 연산이 가능하다
파이썬에서는 글자(문자열)끼리도 연산 기호를 쓸 수 있습니다.
* `+` : 문자열과 문자열을 이어 붙임 (예: `"안녕" + "하세요"` → `안녕하세요`)
* `*` : 문자열을 여러 번 반복함 (예: `"하" * 3` → `하하하`)
* ⚠️ 단, **문자열끼리는 `+`**, **문자열과 숫자는 `*`** 만 가능하고, 문자열과 숫자를 `+`로 연결하면 에러가 납니다! (`"나이는" + 17` ❌)


````python
# [예제] 따옴표 유무의 차이
print("100 + 200")  # 글자 그대로 출력
print(100 + 200)    # 계산되어 300 출력

````

````python
# [예제] 문자열 연산과 쉼표 연결
print("안녕" + "하세요")
print("안녕! " * 5)
print("나의 파이썬 점수는", 100, "점입니다!")
````

## **2. 이스케이프 문자 (출력 꾸미기)**

문자열 안에 특별한 기능을 하는 기호를 넣고 싶을 때 **백슬래시(`\`)** 를 사용합니다. 이런 조합을 **이스케이프 문자**라고 부릅니다.

| 이스케이프 문자 | 기능 |
|---|---|
| `\n` | 줄 바꿈 (Enter 역할) |
| `\t` | 탭 간격 띄우기 (칸 맞추기 역할) |
| `\\` | 백슬래시(`\`) 자체를 출력 |
| `\'` | 작은따옴표(`'`)를 문자열 안에 출력 |
| `\"` | 큰따옴표(`"`)를 문자열 안에 출력 |

> 💡 왜 따옴표를 그냥 못 쓸까요? `"나는 "행복"해"` 처럼 큰따옴표 문자열 안에 큰따옴표가 그대로 들어가면, 파이썬은 문자열이 `"나는"`에서 끝났다고 착각해서 에러가 납니다. 이럴 때 `\"`로 표시해주면 "이건 진짜 글자야"라고 알려줄 수 있어요.

````python
# [예제] 이스케이프 문자 활용
print("안녕하세요.\n반갑습니다.")
print("\n--- 영수증 형식 연습 ---")
print("품목\t가격\t수량")
print("사과\t1000\t5")
print("포도\t3000\t2")


````

````python
# [예제] 따옴표와 백슬래시 출력하기
print("선생님이 말씀하셨다: \"오늘 숙제는 없습니다!\"")
print("경로 표시: C:\\Users\\student")
````

## **3. 변수(Variable)와 자료형(`type`)**

* **변수:** 데이터를 담아두는 **'이름표가 붙은 상자'** 입니다. (예: `money = 1000`)
  * `=`는 수학의 '같다'가 아니라 **"오른쪽 값을 왼쪽 상자에 저장하라"** 는 명령입니다.
* **변수 이름 규칙:**
  * 영어, 숫자, 언더바(`_`)만 사용 가능 (한글도 가능하지만 권장하지 않음)
  * 숫자로 시작 불가 (`1money` ❌ / `money1` ⭕)
  * 띄어쓰기 금지! (여러 단어는 언더바로 연결: `my_age`)
  * 파이썬이 이미 사용하는 단어(`print`, `input` 등)는 변수 이름으로 쓰지 않기

* **대표 자료형 3가지**
  * `str` (문자열): 따옴표로 감싼 글자
  * `int` (정수): 소수점 없는 숫자
  * `float` (실수): 소수점이 있는 숫자

* **`type()` 함수:** 어떤 데이터가 무슨 자료형인지 알려줍니다.



````python
# [예제] 변수 만들기와 재사용
money = 1000
money = money + 500  # 1000 + 500을 계산해서 다시 money 상자에 넣음
print("현재 내 돈은:", money)

print("\n--- 자료형 확인 ---")
word = "안녕"
number = 100
pi = 3.14

print(word, "의 자료형은:", type(word))
print(number, "의 자료형은:", type(number))
print(pi, "의 자료형은:", type(pi))


````

````python
# [예제] 한 줄에 여러 변수 만들기 (심화)
a, b, c = 1, 2, 3
print(a, b, c)
````

## **🚨 5. `input()`의 함정과 형변환 (Type Casting)**

* **중요 원리:** `input()`은 우리가 숫자를 입력해도 무조건 **글자(문자열, str)** 로 취급합니다.
그래서 숫자를 입력받아 바로 더하기 연산을 하면 에러가 발생합니다.

```python
age = input("나이: ")   # 문자열로 저장됨! (예: "17")
print(age + 1)          # ❌ TypeError 발생!
```

* **해결책 (형변환):** 글자를 진짜 숫자로 바꿔주는 명령어가 필요합니다.
  * `int()` : 정수(1, 2, 100...)로 변환
  * `float()` : 실수(1.5, 3.14...)로 변환
  * `str()` : 반대로 숫자를 글자로 변환 (숫자와 문자열을 `+`로 이어붙일 때 필요)

* ⚠️ **주의:** `int("3.5")`처럼 소수점이 있는 글자를 바로 `int()`로 바꾸려 하면 에러가 납니다. 이럴 땐 `float()`을 먼저 쓰거나 `int(float("3.5"))`처럼 사용해야 합니다.

* **소수점 제어:** `f-string`에서는 `{변수:.2f}`, `%` 방식에서는 `%.2f`라고 쓰면 소수점 둘째 자리까지만 출력됩니다.

## **7. 파이썬의 산술 연산자**

기본적인 사칙연산(`+`, `-`, `*`) 외에, 프로그래밍에서 자주 쓰이는 4가지 특별한 연산자
* `/` (나누기): 소수점 끝까지 나눔
* `//` (몫): 정수 몫만 구함. **단가나 수량 계산에 필수!**
* `%` (나머지): 남은 나머지만 구함. **거스름돈이나 짝/홀수 판단에 필수!**
* `**` (거듭제곱): 같은 수를 여러 번 곱함 (예: `2 ** 3` = 8)

### 📌 연산자 우선순위
수학과 마찬가지로 `**` → `*`, `/`, `//`, `%` → `+`, `-` 순서로 계산됩니다. 괄호 `()`를 사용하면 우선순위를 바꿀 수 있습니다.

### 🔥 **심화: 복합 대입 연산자**
변수의 값을 계산해서 다시 자기 자신에게 저장할 때, 더 짧게 쓰는 방법입니다.
```python
score = 10
score += 5   # score = score + 5 와 동일 (결과: 15)
score -= 3   # score = score - 3
score *= 2   # score = score * 2
score //= 4  # score = score // 4
```

````python
print("일반 나누기 (10 / 3) :", 10 / 3)
print("몫만 구하기 (10 // 3) :", 10 // 3)
print("나머지 구하기 (10 % 3) :", 10 % 3)
print("거듭제곱 (2 ** 3) :", 2 ** 3)

# 연산자 우선순위 예시
print("괄호 없이:", 2 + 3 * 4)     # 곱셈 먼저 -> 14
print("괄호 사용:", (2 + 3) * 4)   # 괄호 먼저 -> 20

# 복합 대입 연산자 (심화)
score = 10
score += 5
print("복합 대입 연산자 결과:", score)
````

## **8. 문자열 분리 기능: `.split()`**

입력받은 문자열 뒤에 `.split()`을 붙이면, **'띄어쓰기'** 를 기준으로 데이터를 여러 개로 분리합니다.
괄호 안에 `split(",")`처럼 특정 문자를 넣으면 그 문자를 기준으로 쪼개줍니다.

```python
menu1, menu2 = input("좋아하는 음식 2가지를 띄어쓰기로 입력하세요: ").split()
```
위 코드는 사실 두 단계가 합쳐진 것입니다.
1. `input(...)`으로 문자열을 입력받는다. (예: `"짜장면 짬뽕"`)
2. `.split()`으로 띄어쓰기 기준으로 나눠서 리스트를 만든다. (예: `["짜장면", "짬뽕"]`)
3. 나뉜 값을 `menu1, menu2` 두 변수에 순서대로 저장한다.

⚠️ **주의:** 변수 개수와 분리된 데이터 개수가 다르면 `ValueError`가 발생합니다! (예: 2개 변수인데 3개를 입력하면 에러)

````python
# 1. 띄어쓰기로 분리 (예: 짜장면 짬뽕)
menu1, menu2 = input("좋아하는 음식 2가지를 띄어쓰기로 입력하세요: ").split()
print(f"첫 번째: {menu1}, 두 번째: {menu2}")

# 2. 쉼표로 분리 (예: 사과,포도)
fruit1, fruit2 = input("과일 2개를 쉼표(,)로 구분해서 적어주세요: ").split(",")
print(f"첫 번째: {fruit1}, 두 번째: {fruit2}")
````

## **9. 다중 입력과 일괄 형변환: `map(int, input().split())`**

`split()`으로 분리한 데이터는 여전히 글자(str)입니다.
**`map()`** 함수를 사용하면 분리된 모든 문자열을 한꺼번에 정수(`int`)로 변환할 수 있습니다.

```python
num1, num2 = map(int, input("숫자 2개를 띄어쓰기로 입력하세요: ").split())
```
이 코드는 `input().split()`으로 나온 문자열 리스트의 **각 항목에 `int()`를 하나씩 적용**하는 것입니다. 즉, `split()`과 `int()`를 각각 여러 번 쓰지 않고 한 줄로 처리할 수 있는 편리한 문법입니다.

> 💡 실수(float)로 한꺼번에 변환하고 싶다면 `map(float, ...)`로 바꾸면 됩니다.

````python
# 실행 후 빈칸에 '1000 2000' 입력해 보기
num1, num2 = map(int, input("숫자 2개를 띄어쓰기로 입력하세요: ").split())
print("두 숫자의 합:", num1 + num2)
````

### **[도전 1] 동아리 환영 파티 음식 계산기**
파티 참석자의 수를 입력받아 필요한 치킨(1인 1마리), 콜라(1인 2캔), 감자칩(1인 1개)의 총수량을 출력하는 프로그램을 작성하세요.

````python
#동아리 환영 파티 준비 (파티 참석자 입력받기)
# 힌트: people = int(input("파티 참석 인원은 몇 명인가요? "))
# --- [아래에 코드 작성] ---
people = int(input("파티 참석 인원은 몇 명인가요? "))

# 1인당 치킨 1마리, 콜라 2캔, 감자칩 1개가 필요합니다.
# 사람 수(people)를 활용해 아래 빈칸에 알맞은 계산식을 적어보세요!
chicken, chips, cola = people, people, people*2

# 계산한 결과를 화면에 출력해 봅시다. (f-string 활용)
print(f"필요한 치킨은 {chicken}마리, 콜라는 {cola}캔, 감자칩은 {chips}개입니다.")
````

### **[도전 4] 🎓 나만의 성적표 만들기 (최종 종합 미션)**

아래 순서대로 코드를 작성해서, 배운 내용을 모두 활용한 성적표를 완성해 보세요.

1. 이름을 입력받기
2. 국어, 영어, 수학 점수를 **한 번에** 띄어쓰기로 입력받고 정수로 변환하기 (`map` 활용)
3. 세 과목의 총점과 평균(소수점 첫째 자리까지)을 계산하기
4. `\t`, `\n`, `f-string`, 정렬 서식 등을 자유롭게 활용해서 아래와 같은 느낌으로 출력하기
```
========== 성적표 ==========
이름: 홍길동
국어   90
영어   85
수학   95
------------------------------
총점: 270 / 평균: 90.0
========== ========== ==========
```

````python
# [도전 4] 나만의 성적표 만들기
# 1. 이름 입력받기
name = input("이름: ")

# 2. 세 과목 점수 한 번에 입력받기 (map 활용)
kor, eng, math = map(int, input("점수 3개를 띄어쓰기로 구분해 입력: ").split())

# 3. 총점과 평균 계산하기
plus = kor+eng+math
mean = plus/3

# 4. 성적표 형태로 출력하기

print("=" * 30)
print("           성 적 표")
print("=" * 30)

print(f"이름: {name}")
print(f"국어    {kor}")
print(f"영어    {eng}")
print(f"수학    {math}")
print("------------------------------")
print(f"총점: {plus} / 평균: {mean}")
print("========== ========== ==========")
````
